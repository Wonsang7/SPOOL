# SPOOL — photon-efficient quantitative FLIM

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/Wonsang7/SPOOL/blob/main/SPOOL_demo.ipynb)

**SPOOL** (Spatially Pooled Optical Observation Likelihood) is a training-free Poisson inverse framework that models the microscope point-spread function (PSF) explicitly. Instead of estimating lifetime independently at each detector pixel, SPOOL lets photons detected across neighboring pixels jointly constrain the same source.

### What this demo shows
Using the same forward model and simulation parameters as the manuscript's heterogeneous multi-emitter benchmark, this notebook:

1. simulates a few-photon FLIM acquisition containing 2 ns and 4 ns emitters,
2. reconstructs the **same raw photon counts** with pixel-wise Poisson MLE and SPOOL,
3. compares emitter-center lifetime RMSE, and
4. optionally runs a small photon-efficiency sweep.

> **Quick start:** open in Colab, select a GPU runtime, then choose **Runtime → Run all**.  
> **Recommended runtime:** T4 GPU (`Runtime → Change runtime type → T4 GPU`).  
> The main demo is fully simulated; no experimental data download is required.

## 2. How SPOOL uses spatially distributed photons

For detector pixel $i$ and time bin $t$, the forward model is

$$
Y_{i,t}\sim\mathrm{Poisson}\!\left[
\sum_k D_{k,t}\,(h*A_k)_i+b
\right],
$$

where $A_k$ is the source-space amplitude associated with lifetime basis $k$, $D_{k,t}$ is the corresponding temporal decay basis, $h$ is the microscope PSF, and $b$ is background.

**Pixel-wise MLE** estimates the quantitative parameter independently at each detector pixel.  
**SPOOL** instead estimates the source-space coefficients $A_k$ through the PSF-coupled likelihood, allowing photons distributed over one diffraction-limited footprint to constrain the same underlying source.

The notebook calls the repository's benchmark/reconstruction functions directly rather than reimplementing SPOOL here.


## 3. Build the reconstruction operators

The reconstruction uses the manuscript's **lifetime-agnostic dictionary**: 11 IRF-convolved exponential bases from 1.0 to 5.0 ns in 0.4 ns steps. The true 2 ns and 4 ns lifetimes fall between dictionary nodes.

The optical model uses a pixel-integrated Gaussian PSF with 200 nm FWHM at 50 nm sampling. `C` is the sensitivity image used to normalize the multiplicative update.


```python
bm.D_REC = bm.build_decay_basis(bm.TAU_REC)      # (K=11, T=256)

D_d    = bm.to_dev(bm.D_REC)
Dsum_d = bm.to_dev(bm.D_REC.sum(axis=1))

H = W = fs.IMAGE_SIZE
o = bm.make_otf(np.asarray(fs.PSF), (H, W))

if bm.USE_TORCH:
    import torch
    ones = torch.ones(
        (bm.K_REC, H, W),
        dtype=torch.float64,
        device=bm.DEVICE,
    )
    C = bm.conv_batch(ones, o) * Dsum_d.reshape(bm.K_REC, 1, 1)
else:
    ones = np.ones((bm.K_REC, H, W))
    C = bm.conv_batch(ones, o) * np.asarray(Dsum_d).reshape(bm.K_REC, 1, 1)

print(f"✓ Dictionary: K={bm.K_REC}, {bm.TAU_REC[0]:.1f}–{bm.TAU_REC[-1]:.1f} ns")
print(
    f"✓ PSF: {fs.PSF_FWHM_NM:.0f} nm FWHM "
    f"({fs.PSF_SIGMA_PX:.3f} px sigma at {fs.PIXEL_SIZE_NM} nm/px)"
)
```

## 4. Simulate one few-photon acquisition

The scene contains 15 emitters with 2 ns and 4 ns lifetimes mixed in the same field. Photon detection is Poisson, and the background is scaled at each photon level to preserve the foreground signal-to-background ratio used in the manuscript benchmark.

Use the control below to change the photon budget. The default (`200` photons per emitter) gives roughly **5 detected photons per foreground pixel** for this simulated scene.


```python
#@title Choose the photon budget
photons_per_bead = 200  #@param [50, 100, 200, 400, 800] {type:"raw"}

cfg = {"n_species0": None}      # heterogeneous scene: 2 ns + 4 ns
DEMO_SEED = 2000

bg = bm.background_for_level(photons_per_bead, cfg)
fs.BACKGROUND_RATE = bg

rng = np.random.default_rng(DEMO_SEED)
A_true, coords = fs.generate_scene(
    n_beads=bm.N_BEADS,
    photons_per_bead=photons_per_bead,
    rng=rng,
    n_species0=cfg["n_species0"],
)

species = np.array([np.argmax(A_true[:, y, x]) for (x, y) in coords])
tau_true = np.asarray(fs.TAUS_NS)[species]

Y, Lam = fs.simulate_measurement(A_true, rng)
fg_mean, _ = bm.photon_budget(Lam, coords)

print(f"✓ Simulated {len(coords)} emitters")
print(f"✓ Mean detected photons per foreground pixel: {fg_mean:.1f}")
print(f"✓ Background rate used by the benchmark: {bg:.4g}")
```

## 5. Reconstruct the same photon counts: pixel-wise MLE vs SPOOL

Both reconstructions receive the **same measurement `Y`**, the same 11-component lifetime dictionary, and the same background. The key difference is spatial modeling:

- **Pixel-wise MLE:** each detector pixel is fitted independently.
- **SPOOL:** the PSF is included in the likelihood, coupling neighboring detector pixels to the same source-space estimate.

The SPOOL benchmark uses 50 iterations with η = 0.9, fixed across the manuscript datasets.


```python
t0 = time.time()

with bm.reconstruction_dictionary():
    Y_d = bm.to_dev(Y)
    A_mle = bm.to_np(bm.mle_gpu(Y_d, D_d, Dsum_d, bg))
    A_sp  = bm.to_np(bm.joint_gpu(Y_d, D_d, o, C, bg))

elapsed = time.time() - t0

tau_mle = bm.lifetime_at(A_mle, coords, bm.TAU_REC)
tau_sp  = bm.lifetime_at(A_sp,  coords, bm.TAU_REC)

rmse_mle = float(np.sqrt(np.mean((tau_mle - tau_true) ** 2)))
rmse_sp  = float(np.sqrt(np.mean((tau_sp  - tau_true) ** 2)))
improvement = rmse_mle / rmse_sp if rmse_sp > 0 else np.inf

print(f"✓ Reconstruction time: {elapsed:.1f} s ({bm.BACKEND})")
print()
print("Emitter-center lifetime RMSE")
print(f"  Pixel-wise MLE : {rmse_mle:.3f} ns")
print(f"  SPOOL          : {rmse_sp:.3f} ns")
print(f"  RMSE improvement: {improvement:.2f}×")
```

```python
cmap = plt.cm.rainbow.copy()
cmap.set_bad("black")

panels = [
    (
        "Ground truth",
        bm.masked_map(
            bm.tau_map(A_true, np.asarray(fs.TAUS_NS)),
            A_true.sum(0),
            frac=0.5,
        ),
    ),
    ("Detected photons", None),
    (
        "Pixel-wise MLE",
        bm.masked_map(
            bm.tau_map(A_mle, bm.TAU_REC),
            A_mle.sum(0),
            frac=0.05,
        ),
    ),
    (
        "SPOOL",
        bm.masked_map(
            bm.tau_map(A_sp, bm.TAU_REC),
            A_sp.sum(0),
            frac=0.05,
        ),
    ),
]

fig, axes = plt.subplots(1, 4, figsize=(16, 4.2))

for ax, (title, lifetime_map) in zip(axes, panels):
    if lifetime_map is None:
        im = ax.imshow(Y.sum(axis=2), cmap="magma")
        fig.colorbar(im, ax=ax, fraction=0.046, label="photons")
    else:
        im = ax.imshow(lifetime_map, cmap=cmap, vmin=2, vmax=4)
        fig.colorbar(im, ax=ax, fraction=0.046, label="lifetime (ns)")

    ax.set_title(title)
    ax.axis("off")

plt.suptitle(
    f"Heterogeneous phantom — {fg_mean:.1f} photons per foreground pixel",
    y=1.03,
)
plt.tight_layout()
plt.show()
```

> **Key observation:** the pixel-wise reconstruction estimates lifetime separately across the diffraction-limited footprint, whereas SPOOL uses the spatial photon distribution through the PSF-coupled likelihood. The emitter-center RMSE printed above provides the quantitative comparison for this trial.


## 6. Optional: miniature photon-efficiency sweep

This quick sweep uses three photon levels × three Poisson trials. It is intended as an interactive Colab demonstration, not as a replacement for the full manuscript benchmark.

It is **off by default** so that `Runtime → Run all` finishes after the core single-trial demonstration. Turn it on below if you want to reproduce the trend across photon levels.

For the full benchmark (five photon levels × eight trials plus all baselines), run `python run_multiemitter_benchmark.py`.


```python
#@title Run the miniature photon sweep?
RUN_PHOTON_SWEEP = False  #@param {type:"boolean"}

if not RUN_PHOTON_SWEEP:
    print("Mini sweep skipped. Enable the checkbox and rerun this cell to run it.")
else:
    levels, n_trials = [50, 200, 800], 3
    sweep = {"MLE": [], "SPOOL": []}
    fg_x = []

    # Preserve the main-demo background setting so this optional cell has
    # no side effects on later interactive use.
    background_before_sweep = fs.BACKGROUND_RATE

    try:
        for pl in levels:
            bg_sweep = bm.background_for_level(pl, cfg)
            fs.BACKGROUND_RATE = bg_sweep

            r_m, r_s, f_ = [], [], []

            for trial in range(n_trials):
                rng = np.random.default_rng(DEMO_SEED + trial)

                A_t, crd = fs.generate_scene(
                    n_beads=bm.N_BEADS,
                    photons_per_bead=pl,
                    rng=rng,
                    n_species0=cfg["n_species0"],
                )

                sp = np.array([np.argmax(A_t[:, y, x]) for (x, y) in crd])
                t_t = np.asarray(fs.TAUS_NS)[sp]

                Yt, Lt = fs.simulate_measurement(A_t, rng)
                f_.append(bm.photon_budget(Lt, crd)[0])

                with bm.reconstruction_dictionary():
                    Yd = bm.to_dev(Yt)
                    Am = bm.to_np(bm.mle_gpu(Yd, D_d, Dsum_d, bg_sweep))
                    Aj = bm.to_np(bm.joint_gpu(Yd, D_d, o, C, bg_sweep))

                r_m.append(
                    np.sqrt(
                        np.mean(
                            (bm.lifetime_at(Am, crd, bm.TAU_REC) - t_t) ** 2
                        )
                    )
                )
                r_s.append(
                    np.sqrt(
                        np.mean(
                            (bm.lifetime_at(Aj, crd, bm.TAU_REC) - t_t) ** 2
                        )
                    )
                )

            sweep["MLE"].append((np.mean(r_m), np.std(r_m)))
            sweep["SPOOL"].append((np.mean(r_s), np.std(r_s)))
            fg_x.append(np.mean(f_))

            print(
                f"{fg_x[-1]:5.1f} photons/px | "
                f"MLE {np.mean(r_m):.3f} ns | "
                f"SPOOL {np.mean(r_s):.3f} ns"
            )
    finally:
        fs.BACKGROUND_RATE = background_before_sweep

    plt.figure(figsize=(5.5, 4))

    for name, color in [("MLE", "gray"), ("SPOOL", "crimson")]:
        mu = [v[0] for v in sweep[name]]
        sd = [v[1] for v in sweep[name]]
        plt.errorbar(
            fg_x,
            mu,
            yerr=sd,
            marker="o",
            color=color,
            label=name,
            capsize=3,
        )

    plt.xscale("log")
    plt.xlabel("mean photons per foreground pixel")
    plt.ylabel("emitter-center lifetime RMSE (ns)")
    plt.title("Heterogeneous phantom — mini sweep")
    plt.legend()
    plt.tight_layout()
    plt.show()
```
